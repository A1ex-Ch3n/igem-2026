```python
import pandas as pd
st = pd.read_csv('root.state', sep='\t', comment='#')
##### Ancestral nodes:  1, 2, 38,39,40,42,267,277,281
##### The ancestral nodes above are at midpoint rooting
# nodes = ['Node1', 'Node2', 'Node38', 'Node39', 'Node40', 'Node42', 'Node267', 'Node277', 'Node281']
with open('all_nodes.fa', 'w') as f:

  for i in range(1,999):
    node = 'Node'+ str(i)
    #for node in nodes:
    sub = st[st.Node == node].sort_values('Site')
    seq = ''.join(sub.State)
    f.write('>'+ node + '\n' + seq + '\n')
```

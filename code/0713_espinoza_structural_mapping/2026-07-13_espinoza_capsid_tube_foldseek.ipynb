```python
from Bio.PDB import PDBList, MMCIFParser

pdbl = PDBList()
filename = pdbl.retrieve_pdb_file("8h89", file_format="mmCif")

parser = MMCIFParser(QUIET=True)
structure = parser.get_structure("8h89", filename)

for chain in structure[0]:
    print(chain.id)
```

```python
import py3Dmol
# Replace '1ycr' with your local pdb file path
view = py3Dmol.view(query='pdb:8jov')
# Everything else in gray
#view.setStyle(
#    {'chain': ['C','D','E','F','G','H']},
#    {'cartoon': {'color': 'lightgray', 'opacity': 0.3}}
#)

# CDS_20 Portal Protein
view.setStyle(
    {'chain': '0'},
    {'cartoon': {'color': 'red'}}
)
# CDS_008
view.setStyle(
    {'chain': ['A', 'B', 'F']},
    {'cartoon': {'color': 'Black'}}
)

# CDS 09
view.setStyle(
    {'chain': ['1', '4', '7']},
    {'cartoon': {'color': 'blue'}}
)
# CDS_013
view.setStyle(
    {'chain': ['G', 'H', 'I']},
    {'cartoon': {'color': 'yellow'}}
)
# CDS_014
view.setStyle(
    {'chain': 'T'},
    {'cartoon': {'color': 'green'}}
)







view.zoomTo()
view.show()
```
